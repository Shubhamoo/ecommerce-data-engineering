```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
spark = SparkSession.builder.appName("EcomDataPipeline").getOrCreate()
```

```python
userDF = (
    spark.read
    .format("parquet")
    .load("abfss://landing-zone-2@ecomadls.dfs.core.windows.net/users-raw-2/")
)

userDF.show(5)
```

```python
userDF = spark.read.parquet(
    "abfss://landing-zone-2@ecomadls.dfs.core.windows.net/users-raw-2/"
)

buyerDF = spark.read.parquet(
    "abfss://landing-zone-2@ecomadls.dfs.core.windows.net/buyers-raw-2/"
)

sellerDF = spark.read.parquet(
    "abfss://landing-zone-2@ecomadls.dfs.core.windows.net/sellers-raw-2/"
)

countryDF = spark.read.parquet(
    "abfss://landing-zone-2@ecomadls.dfs.core.windows.net/countries-raw-2/"
)
```

```python
userDF.write.format("delta") \
    .mode("overwrite")\
    .save("abfss://landing-zone-2@ecomadls.dfs.core.windows.net/users")
```

```python
buyerDF.write.format("delta") \
    .mode("overwrite")\
    .save("abfss://landing-zone-2@ecomadls.dfs.core.windows.net/buyers")
```

```python
sellerDF.write.format("delta") \
    .mode("overwrite")\
    .save("abfss://landing-zone-2@ecomadls.dfs.core.windows.net/sellers")
```

```python
countryDF.write.format("delta") \
    .mode("overwrite")\
    .save("abfss://landing-zone-2@ecomadls.dfs.core.windows.net/countries")
```
